```python
seconds_per_hour = 60 * 60
print("Seconds in a hour: ", seconds_per_hour)
seconds_per_day = seconds_per_hour * 24
print("Seconds in a day: ", seconds_per_day)

answer5 = seconds_per_day / seconds_per_hour
print("answer5 ", answer5)
answer6 = seconds_per_day // seconds_per_hour
print("answer6 ", answer6)
```
